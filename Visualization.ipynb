```python
import numpy as np
import matplotlib.pyplot as plt
import os

def plot_icm_stats(npy_path, save_dir="results/icm/plots"):
    # ---- Load ----
    data = np.load(npy_path, allow_pickle=True).item()
    episode_rewards = data["episode_rewards"]
    episode_intrinsic = data["episode_intrinsic"]
    episode_lengths = data["episode_lengths"]
    rewarded_steps = data["Total Rewards"]
    all_step_rewards = data["all_step_rewards"]

    os.makedirs(save_dir, exist_ok=True)

    # ---- Plot 1: Episode Rewards ----
    plt.figure(figsize=(6,4))
    plt.plot(episode_rewards, marker="o")
    plt.title("Episode External Rewards")
    plt.xlabel("Episode")
    plt.ylabel("Reward")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(f"{save_dir}/episode_rewards.png")
    plt.show()

    # ---- Plot 2: Episode Intrinsic Reward ----
    plt.figure(figsize=(6,4))
    plt.plot(episode_intrinsic, marker="o", color="orange")
    plt.title("Episode Intrinsic Rewards (ICM)")
    plt.xlabel("Episode")
    plt.ylabel("Intrinsic Reward")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(f"{save_dir}/episode_intrinsic.png")
    plt.show()


    # ---- Plot 3: Rewarded Step Curve ----
    plt.figure(figsize=(6,4))
    plt.plot(rewarded_steps)
    plt.title("all step reward")
    plt.xlabel("episode")
    plt.ylabel("Total Reward")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(f"{save_dir}/all_rewarded_steps.png")
    plt.show()

    print(f"[INFO] All plots saved to: {save_dir}")

    # ---- Plot 4: Episode Lengths ----
    plt.figure(figsize=(6,4))
    plt.plot(episode_lengths, marker="o", color="green")
    plt.title("Episode Lengths")
    plt.xlabel("Episode")
    plt.ylabel("Length")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(f"{save_dir}/episode_lengths.png")
    plt.show()

    print(f"[INFO] All plots saved to: {save_dir}")


if __name__ == "__main__":
    # ===== 修改这里的路径 =====
    npy_path = "results/icm/minigrid_icm_stats_20251129_225923.npy"
    plot_icm_stats(npy_path)
```
