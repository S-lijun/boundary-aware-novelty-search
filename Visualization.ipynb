```python
import numpy as np
import matplotlib.pyplot as plt
import os

def plot_icm_stats(npy_path, save_dir="results/icm/plots"):
    # ---- Load ----
    data = np.load(npy_path, allow_pickle=True).item()
    episode_rewards = data["episode_rewards"]
    episode_intrinsic = data["episode_intrinsic"]
    episode_lengths = data["episode_lengths"]
    rewarded_steps = data["Total Rewards"]
    all_step_rewards = data["all_step_rewards"]

    os.makedirs(save_dir, exist_ok=True)

    # ---- Plot 1: Episode Rewards ----
    plt.figure(figsize=(6,4))
    plt.plot(episode_rewards, marker="o")
    plt.title("Episode External Rewards")
    plt.xlabel("Episode")
    plt.ylabel("Reward")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(f"{save_dir}/episode_rewards.png")
    plt.show()

    # ---- Plot 2: Episode Intrinsic Reward ----
    plt.figure(figsize=(6,4))
    plt.plot(episode_intrinsic, marker="o", color="orange")
    plt.title("Episode Intrinsic Rewards (ICM)")
    plt.xlabel("Episode")
    plt.ylabel("Intrinsic Reward")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(f"{save_dir}/episode_intrinsic.png")
    plt.show()


    # ---- Plot 3: Rewarded Step Curve ----
    plt.figure(figsize=(6,4))
    plt.plot(rewarded_steps)
    plt.title("all step reward")
    plt.xlabel("episode")
    plt.ylabel("Total Reward")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(f"{save_dir}/all_rewarded_steps.png")
    plt.show()

    print(f"[INFO] All plots saved to: {save_dir}")

    # ---- Plot 4: Episode Lengths ----
    plt.figure(figsize=(6,4))
    plt.plot(episode_lengths, marker="o", color="green")
    plt.title("Episode Lengths")
    plt.xlabel("Episode")
    plt.ylabel("Length")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(f"{save_dir}/episode_lengths.png")
    plt.show()

    print(f"[INFO] All plots saved to: {save_dir}")


if __name__ == "__main__":
    # ===== 修改这里的路径 =====
    npy_path = "results/icm/minigrid_icm_stats_20251130_030006.npy"
    plot_icm_stats(npy_path)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os

# ---- 加一个漂亮的style ----
plt.style.use("seaborn-v0_8")

def plot_icm_stats(npy_path, save_dir="results/icm/plots"):
    data = np.load(npy_path, allow_pickle=True).item()
    episode_rewards = data["episode_rewards"]
    episode_intrinsic = data["episode_intrinsic"]
    episode_lengths = data["episode_lengths"]
    rewarded_steps = data["Total Rewards"]

    os.makedirs(save_dir, exist_ok=True)

    # ---- Plot 1: Episode Rewards ----
    plt.figure(figsize=(6,4))
    plt.plot(episode_rewards, marker="o", markersize=2, linewidth=0.7)
    plt.title("Episode External Rewards")
    plt.xlabel("Episode")
    plt.ylabel("Reward")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(f"{save_dir}/episode_rewards.png")
    plt.show()

    # ---- Plot 2: Episode Intrinsic Reward ----
    plt.figure(figsize=(6,4))
    plt.plot(episode_intrinsic, marker="o", color="orange", markersize=2, linewidth=0.7)
    plt.title("Episode Intrinsic Rewards (ICM)")
    plt.xlabel("Episode")
    plt.ylabel("Intrinsic Reward")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(f"{save_dir}/episode_intrinsic.png")
    plt.show()

    # ---- Plot 3: Rewarded Step Curve ----
    plt.figure(figsize=(6,4))
    plt.plot(rewarded_steps, markersize=2, linewidth=0.7)
    plt.title("All Step Reward")
    plt.xlabel("Episode")
    plt.ylabel("Total Reward")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(f"{save_dir}/all_rewarded_steps.png")
    plt.show()

    # ---- Plot 4: Episode Lengths ----
    plt.figure(figsize=(6,4))
    plt.plot(episode_lengths, marker="o", color="green", markersize=2, linewidth=0.7)
    plt.title("Episode Lengths")
    plt.xlabel("Episode")
    plt.ylabel("Length")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(f"{save_dir}/episode_lengths.png")
    plt.show()

    print(f"[INFO] All plots saved to: {save_dir}")

if __name__ == "__main__":
    # ===== 修改这里的路径 =====
    npy_path = "results/icm/minigrid_icm_stats_20251130_030006.npy"
    npy_path = "results/rnd/minigrid_rnd_stats_20251130_184429.npy"
    plot_icm_stats(npy_path)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os

plt.style.use("seaborn-v0_8")


def plot_icm_compare(paths, labels=None, save_dir="results/compare_plots"):
    os.makedirs(save_dir, exist_ok=True)

    if labels is None:
        labels = [f"Model {i}" for i in range(len(paths))]

    all_data = []

    # ------------ Load all .npy ------------
    for p in paths:
        d = np.load(p, allow_pickle=True).item()

        episode_rewards = d["episode_rewards"]
        episode_intrinsic = d["episode_intrinsic"]
        episode_lengths = d["episode_lengths"]

        # step reward (raw)
        if "all_step_rewards" in d:
            step_rewards = d["all_step_rewards"]
        elif "rewarded_steps" in d:
            step_rewards = d["rewarded_steps"]
        elif "Total Rewards" in d:
            step_rewards = d["Total Rewards"]
        else:
            raise ValueError("No step reward found in file!")

        # ------------ Convert to TOTAL REWARD (CUMULATIVE) ------------
        total_reward_curve = np.cumsum(step_rewards)

        all_data.append({
            "episode_rewards": episode_rewards,
            "episode_intrinsic": episode_intrinsic,
            "episode_lengths": episode_lengths,
            "total_reward_curve": total_reward_curve
        })

    # ============================================================
    # Plot 1: Episode External Rewards
    # ============================================================
    plt.figure(figsize=(7, 5))
    for i, d in enumerate(all_data):
        plt.plot(d["episode_rewards"], label=labels[i], linewidth=1)
    plt.title("Episode External Rewards")
    plt.xlabel("Episode")
    plt.ylabel("Reward")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig(f"{save_dir}/compare_episode_rewards.png")
    plt.show()

    # ============================================================
    # Plot 2: Episode Intrinsic Rewards
    # ============================================================
    plt.figure(figsize=(7, 5))
    for i, d in enumerate(all_data):
        plt.plot(d["episode_intrinsic"], label=labels[i], linewidth=1)
    plt.title("Episode Intrinsic Rewards")
    plt.xlabel("Episode")
    plt.ylabel("Intrinsic Reward")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig(f"{save_dir}/compare_episode_intrinsic.png")
    plt.show()

    # ============================================================
    # Plot 3: Total Reward (Cumulative)
    # ============================================================
    plt.figure(figsize=(7, 5))
    for i, d in enumerate(all_data):
        plt.plot(d["total_reward_curve"], label=labels[i], linewidth=1)
    plt.title("Total Reward Curve (Cumulative)")
    plt.xlabel("Step")
    plt.ylabel("Total Reward")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig(f"{save_dir}/compare_total_reward_curve.png")
    plt.show()

    # ============================================================
    # Plot 4: Episode Lengths
    # ============================================================
    plt.figure(figsize=(7, 5))
    for i, d in enumerate(all_data):
        plt.plot(d["episode_lengths"], label=labels[i], linewidth=1)
    plt.title("Episode Lengths")
    plt.xlabel("Episode")
    plt.ylabel("Length")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig(f"{save_dir}/compare_episode_lengths.png")
    plt.show()

    print(f"[INFO] Comparison plots saved to: {save_dir}")


if __name__ == "__main__":
    paths = [
        "results/icm/minigrid_icm_stats_20251201_150231.npy",
        "results/rnd/minigrid_rnd_stats_20251201_153128.npy",
        "results/ppo/minigrid_ppo_stats_20251201_180210.npy"
    ]
    labels = ["PPO + ICM", "PPO + RND", "PPO"]

    plot_icm_compare(paths, labels)
```
